Lecture des données SWC choisies (ex: Cortex, pyramidal vs interneurone, mutant vs contrôle), déposer les fichiers dans /data/raw/…, et créer un manifest CSV (chemin, type, condition).

```python
from pathlib import Path
import pandas as pd

RAW = Path("../data/external/swain")
files = list(RAW.rglob("*.swc"))
meta = []
for f in files:
    meta.append({"path": str(f), "morphotype": "pyramidal", "condition": "control"})
df = pd.DataFrame(meta)
df.to_csv("../data/interim/manifest.csv", index=False)
df.head()
```
