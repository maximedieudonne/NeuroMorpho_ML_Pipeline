```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import umap

X = pd.read_csv("../data/processed/features.csv")
y = X[["morphotype","condition"]]; Xn = X.drop(columns=["path","morphotype","condition"]).fillna(0)
Xz = StandardScaler().fit_transform(Xn)

pca = PCA(n_components=2).fit_transform(Xz)
tsne = TSNE(n_components=2, init="pca", learning_rate="auto").fit_transform(Xz)
um = umap.UMAP(n_components=2).fit_transform(Xz)
# scatter plots avec matplotlib (3 figures)
```
