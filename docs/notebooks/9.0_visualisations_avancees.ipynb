```python
import numpy as np, matplotlib.pyplot as plt, neurom as nm, pandas as pd
# Exemple density map simple:
all_pts = []
for p in pd.read_csv("../data/interim/manifest.csv")["path"]:
    n = nm.load_neuron(p)
    pts = np.vstack([s.points[:,:3] for s in n.iter_sections()])
    all_pts.append(pts)
P = np.vstack(all_pts)
H, xedges, yedges = np.histogram2d(P[:,0], P[:,1], bins=200)
plt.imshow(H.T, origin="lower"); plt.title("Carte de densité XY")
```
