```python
import neurom as nm
from neurom import load_neuron, load_neurons
from neurom.core.types import NeuriteType
import pandas as pd, numpy as np
from pathlib import Path

manifest = pd.read_csv("../data/interim/manifest.csv")
nrns = [load_neuron(p) for p in manifest["path"]]

def qc_basic(n):
    return {
        "n_sections": nm.get("number_of_sections", n),
        "n_bifurcations": nm.get("number_of_bifurcations", n),
        "has_zero_diam": any(seg[2]==0 for seg in nm.iter_segments(n))
    }

qc = pd.DataFrame([qc_basic(n) for n in nrns])
qc.to_csv("../data/interim/qc.csv", index=False)
qc.describe()
```
